```python
sorted_data = sorted(sum_purchases)
n = len(sorted_data)

if n % 2 == 1:
    median_python = sorted_data[n // 2]
else:
    median_python = (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2
print("Медіана python:", median_python)

median_numpy = np.median(sum_purchases)
print("Медіана np:", median_numpy)
```

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
mean_x = sum(sum_purchases) / (len(sum_purchases))
mean_y = sum(visits) / (len(visits))
xy = [ (sum_purchases[i] - mean_x)*(visits[i] - mean_y) for i in range(len(sum_purchases))]
cov_xy = sum(xy)/(len(sum_purchases)-1)
variance_x_sample=sum((x - mean_x) ** 2 for x in sum_purchases) / (len(sum_purchases) - 1)
variance_y_sample=sum((x - mean_y) ** 2 for x in visits) / (len(visits) - 1)
std_x_sample=variance_x_sample**0.5
std_y_sample=variance_y_sample**0.5

pearson_python=cov_xy / (std_x_sample * std_y_sample)

print("Коефіцієнт кореляції Пірсона (python):", pearson_python)


pearson_np=np.corrcoef(sum_purchases,visits)[0, 1]
print("Коефіцієнт кореляції Пірсона (np):", pearson_np)

print("Результат говорить про слабку від'ємну кореляцію")
```

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
sum_purchases = np.array(sum_purchases)
visits = np.array(visits)
mean_purchase = np.mean(sum_purchases)
std_purchase = np.std(sum_purchases)
threshold = mean_purchase + 2 * std_purchase
print("Threshold:", threshold)

mask_inliers = sum_purchases <= threshold
sum_purchases_wo_outliers = sum_purchases[mask_inliers]
visits_wo_outliers = visits[mask_inliers]

print("Кількість записів після очищення:", len(sum_purchases_wo_outliers))


pearson_wo_outliers = np.corrcoef(sum_purchases_wo_outliers, visits_wo_outliers)[0, 1]

print("Коефіцієнт кореляції Пірсона без викидів:", pearson_wo_outliers)
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases_wo_outliers, visits_wo_outliers)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

**Висновок**
Після очищення даних лінійної залежності не виявлено, кореляція слабка, вираженого тренду не спостерігається. Відтак, по кількості візитів ми не можемо реально спрогнозувати суму покупок. Імовірно тут не лінійна залежність

```python
from scipy.stats import spearmanr
spearman_corr, p_value = spearmanr(sum_purchases_wo_outliers, visits_wo_outliers)

print("Spearman correlation:", spearman_corr)
print("p-value:", p_value)
```
