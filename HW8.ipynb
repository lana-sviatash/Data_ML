**Про датасет:**

Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
mean_bill = df['total_bill'].mean()
max_tip = df['tip'].max()
isnull_count = df.isna().sum()
dinner_lunch_bill = df.groupby('time').size()

print(f'Середній розмір рахунку на всьому наборі даних: {mean_bill}\n')
print(f'Максимальні чайові: {max_tip}\n')
print(f'Кількість пропущених значень у всіх колонках: \n{isnull_count}\n')
print(f'Скільки в наборі даних рахунків за обіди і за вечері: \n{dinner_lunch_bill}')

# df['time'].unique()
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
bill_by_sex = df.groupby('sex')['total_bill'].mean()
tip_by_day = df.groupby('day')['tip'].mean()
visitor_by_day = df.groupby('day')['size'].sum()

print(f'Середній рахунок за статтю: \n{bill_by_sex}\n')
print(f'Середні чайові за днем тижня: \n{tip_by_day}\n')
print(f'Скільки відвідувачів у кожен день: \n{visitor_by_day}\n')
```

**6. Створення нової колонки**

Створіть нову колонку `tip_percentage` - відсоток чайових від загальної суми рахунку.
Виведіть перші 10 рядків з новою колонкою.

```python
df['tip_percentage'] = (df['tip'] / df['total_bill']).round(4)
df['tip_percentage, %'] = (df['tip'] / df['total_bill'] *100).round(2)
df.head(10)
```
