**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var1 = 15
var2 = 4

print(f"Сума: {var1 + var2}")
print(f"Різниця: {var1 - var2}")
print(f"Добуток: {var1 * var2}")
print(f"Ділення: {var1 / var2}")
print(f"Цілочисельне ділення: {var1 // var2}")
print(f"Залишок від ділення: {var1%var2}")
print(f"Дробовий залишок від ділення: {var1/var2 - var1//var2}")
print(f"Піднесення до степеня: {var1 ** var2}")
```
